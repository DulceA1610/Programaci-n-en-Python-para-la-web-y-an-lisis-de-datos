##**4. Scikit-learn**


*   ¿Qué es y para qué sirve?

Es una de estas librerías gratuitas para Python. Cuenta con algoritmos de clasificación, regresión, clustering y reducción de dimensionalidad, los algoritmos de Scikit-Learn se combinan y depuran con otras estructuras de datos y aplicaciones externas como Pandas o PyBrain, su ventaja es la variedad de módulos y algoritmos que facilitan el aprendizaje y trabajo del científico de datos en las primeras fases de su desarrollo.

*   Principales aplicaciones en aprendizaje automático

Se dividen en cuatro áreas clave:

Clasificación: Asignar categorías a datos, como la detección de spam en correos electrónicos, el reconocimiento de imágenes o la evaluación de riesgos financieros.

Regresión: Predecir valores numéricos continuos, utilizado comúnmente para la predicción de precios, ventas futuras o tendencias de mercado.

Agrupamiento (Clustering): Identificar patrones ocultos en datos no etiquetados, fundamental para la segmentación de clientes y la detección de anomalías.

Reducción de dimensionalidad: Simplificar conjuntos de datos complejos mediante técnicas como PCA, facilitando la visualización y mejorando la eficiencia computacional.

Scikit-Learn, herramienta básica para el Data Science en Python. (2018, August 6). Máster En Data Science. https://www.master-data-scientist.com/scikit-learn-data-science/
‌
*   Ejemplo de implementación
El ejemplo predice si un empleado terminará su relación laboral con la empresa, a partir de características de su historial laboral.


```python

# 1. Importamos librerías

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    ConfusionMatrixDisplay,
    roc_auc_score
)
# 2. Copirar y reparar los datos
datos = df.copy()

# Eliminar espacios en los nombres de las columnas.
datos.columns = datos.columns.str.strip()

# Verificar que exista la variable objetivo.
if "Termd" not in datos.columns:
    raise ValueError("No existe la columna Termd en el DataFrame.")

# Convertir la variable objetivo a valores numéricos.
datos["Termd"] = pd.to_numeric(datos["Termd"], errors="coerce")

# Conservar únicamente registros con objetivo válido.
datos = datos.dropna(subset=["Termd"]).copy()

# Verificar que la variable objetivo sea binaria.
if not set(datos["Termd"].unique()).issubset({0, 1}):
    raise ValueError("Termd debe contener únicamente 0 y 1.")

datos["Termd"] = datos["Termd"].astype(int)
# 3. crear variables de fechas

fecha_ingreso = pd.to_datetime(
    datos["DateofHire"],
    errors="coerce"
)

datos["HireYear"] = fecha_ingreso.dt.year
datos["HireMonth"] = fecha_ingreso.dt.month

# 4. definición de variables de entrada y salida
# Variables numéricas relacionadas con el trabajo.
variables_numericas = [
    "Salary",
    "EngagementSurvey",
    "EmpSatisfaction",
    "SpecialProjectsCount",
    "DaysLateLast30",
    "Absences",
    "HireYear",
    "HireMonth"
]

# Variables categóricas relacionadas con el trabajo.
variables_categoricas = [
    "Department",
    "Position",
    "RecruitmentSource",
    "PerformanceScore"
]

# Comprobar que existan todas las columnas necesarias.
variables = variables_numericas + variables_categoricas

faltantes = [c for c in variables if c not in datos.columns]

if faltantes:
    raise ValueError(f"Faltan estas columnas: {faltantes}")

# X = información que utilizará el modelo.
X = datos[variables].copy()

# y = resultado que queremos aprender a predecir.
y = datos["Termd"]

print("Dimensiones de X:", X.shape)
print("\nDistribución de la variable objetivo:")
print(y.value_counts())
print("\nProporciones de la variable objetivo:")
print(y.value_counts(normalize=True).round(3))
# 5. División de los datos entrenamiento y prueba

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nRegistros para entrenar:", len(X_train))
print("Registros para evaluar:", len(X_test))
# 6. preprocesar variables
# Variables numéricas:
# reemplazar valores faltantes por la mediana.
pipeline_numerico = Pipeline(steps=[
    ("imputador", SimpleImputer(strategy="median"))
])

# Variables categóricas:
# reemplazar faltantes por la categoría más frecuente
# y convertir categorías en columnas numéricas.
pipeline_categorico = Pipeline(steps=[
    ("imputador", SimpleImputer(strategy="most_frequent")),
    ("codificador", OneHotEncoder(handle_unknown="ignore"))
])

preprocesador = ColumnTransformer(transformers=[
    ("numericas", pipeline_numerico, variables_numericas),
    ("categoricas", pipeline_categorico, variables_categoricas)
])
# 7. crear el modelo

modelo = Pipeline(steps=[
    ("preprocesamiento", preprocesador),
    ("clasificador", RandomForestClassifier(
        n_estimators=300,
        max_depth=5,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42
    ))
])

# 8. entrenamos modelo

modelo.fit(X_train, y_train)

print("\nModelo entrenado correctamente.")
# 9. Realizamos predicción

predicciones = modelo.predict(X_test)

# Probabilidad estimada para la clase 1 (Termd = 1).
probabilidades = modelo.predict_proba(X_test)[:, 1]
# 10. Evaluamos rendimiento

accuracy = accuracy_score(y_test, predicciones)

precision = precision_score(
    y_test, predicciones, zero_division=0
)

recall = recall_score(
    y_test, predicciones, zero_division=0
)

f1 = f1_score(
    y_test, predicciones, zero_division=0
)

print("\n========== MÉTRICAS DEL MODELO ==========")
print(f"Accuracy:  {accuracy:.3f}")
print(f"Precision: {precision:.3f}")
print(f"Recall:    {recall:.3f}")
print(f"F1-score:  {f1:.3f}")

print("\nReporte de clasificación:")
print(classification_report(
    y_test,
    predicciones,
    labels=[0, 1],
    target_names=[
        "No terminó su relación laboral",
        "Terminó su relación laboral"
    ],
    zero_division=0
))

if len(np.unique(y_test)) == 2:
    auc = roc_auc_score(y_test, probabilidades)
    print(f"ROC-AUC:   {auc:.3f}")

# 11. matrix de confusión

ConfusionMatrixDisplay.from_predictions(
    y_test,
    predicciones,
    labels=[0, 1],
    display_labels=["No terminó", "Terminó"],
    cmap="Blues",
    values_format="d"
)

plt.title("Matriz de confusión - Predicción de rotación")
plt.tight_layout()
plt.show()

#12. identificar variables importantes

preprocesador_entrenado = modelo.named_steps["preprocesamiento"]

nombres_variables = preprocesador_entrenado.get_feature_names_out()

importancias = modelo.named_steps[
    "clasificador"
].feature_importances_

tabla_importancias = pd.DataFrame({
    "Variable": nombres_variables,
    "Importancia": importancias
})

tabla_importancias = tabla_importancias.sort_values(
    by="Importancia",
    ascending=False
)

print("\n========== VARIABLES MÁS IMPORTANTES ==========")
print(tabla_importancias.head(15).to_string(index=False))

# Gráfica de las 15 variables más importantes.
top_variables = tabla_importancias.head(15).sort_values(
    "Importancia",
    ascending=True
)

plt.figure(figsize=(10, 7))

plt.barh(
    top_variables["Variable"],
    top_variables["Importancia"]
)

plt.title("Variables más importantes para el modelo")
plt.xlabel("Importancia relativa")
plt.ylabel("Variable")
plt.tight_layout()
plt.show()

#comparamos resultados reales vs predichos
resultados = pd.DataFrame({
    "Resultado_real": y_test,
    "Prediccion": predicciones,
    "Probabilidad_estimada_Termd_1": probabilidades
}, index=X_test.index)

print("\nEjemplo de predicciones")
print(resultados.head(10))
```

##**5. TensorFlow**


*   ¿Qué es y para qué se utiliza?

Es una biblioteca de código abierto para la computación numérica, el aprendizaje automático a gran escala, el aprendizaje profundo y otras cargas de trabajo de análisis predictivo y estadístico. Este tipo de tecnología hace que sea más rápido y fácil para los desarrolladores implementar modelos de aprendizaje automático, ya que ayuda en el proceso de adquisición de datos, la entrega de predicciones a escala y el refinamiento de los resultados futuros.

*   Aplicaciones en aprendizaje profundo

Redes Neuronales Convolucionales (CNN): Utilizadas para reconocimiento de imágenes, visión por computadora y análisis de datos médicos, aprovechando la capacidad de TensorFlow para procesar datos espaciales y video.

Procesamiento del Lenguaje Natural (PLN): Permite la creación de modelos para traducción automática, análisis de sentimiento y generación de texto mediante el uso de redes neuronales recurrentes (RNN) y transformers.

Redes Neuronales Recurrentes (RNN): Esencial para tareas secuenciales como reconocimiento de voz y predicción de series temporales.

Implementación y Despliegue: Facilita el entrenamiento de modelos complejos y su posterior despliegue en producción, desde servidores en la nube hasta dispositivos móviles (vía TensorFlow Lite) y navegadores web (vía TensorFlow.js).

*   Ejemplo de construcción de una red neuronal

```python
# 1. IMPORTAR LIBRERÍAS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
from sklearn.utils.class_weight import compute_class_weight

# Fijar semillas para facilitar la reproducibilidad.
np.random.seed(42)
tf.random.set_seed(42)
# 2. PREPARAR EL DATAFRAME
datos = df.copy()
datos.columns = datos.columns.str.strip()

# Verificar la variable que queremos predecir.
if "Termd" not in datos.columns:
    raise ValueError("No se encontró la columna Termd.")

datos["Termd"] = pd.to_numeric(
    datos["Termd"], errors="coerce"
)

datos = datos.dropna(subset=["Termd"]).copy()

if not set(datos["Termd"].unique()).issubset({0, 1}):
    raise ValueError("Termd debe contener únicamente 0 y 1.")

datos["Termd"] = datos["Termd"].astype(int)
# 3. CREAR VARIABLES A PARTIR DE LA FECHA DE INGRESO
fecha_ingreso = pd.to_datetime(
    datos["DateofHire"], errors="coerce"
)

datos["HireYear"] = fecha_ingreso.dt.year
datos["HireMonth"] = fecha_ingreso.dt.month
# 4. DEFINIR LAS VARIABLES PREDICTORAS
variables_numericas = [
    "Salary",
    "EngagementSurvey",
    "EmpSatisfaction",
    "SpecialProjectsCount",
    "DaysLateLast30",
    "Absences",
    "HireYear",
    "HireMonth"
]

variables_categoricas = [
    "Department",
    "Position",
    "RecruitmentSource",
    "PerformanceScore"
]

variables = variables_numericas + variables_categoricas

faltantes = [
    columna for columna in variables
    if columna not in datos.columns
]

if faltantes:
    raise ValueError(f"Faltan columnas: {faltantes}")

X = datos[variables].copy()
y = datos["Termd"].copy()

print("Número de empleados:", len(datos))
print("\nDistribución de Termd:")
print(y.value_counts())
print("\nProporción por clase:")
print(y.value_counts(normalize=True).round(3))
# 5. SEPARAR ENTRENAMIENTO Y PRUEBA
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nRegistros de entrenamiento:", len(X_train))
print("Registros de prueba:", len(X_test))
# 6. PREPROCESAR LOS DATOS
# Numéricas: completar valores faltantes y estandarizar.
pipeline_numerico = Pipeline(steps=[
    ("imputador", SimpleImputer(strategy="median")),
    ("escalador", StandardScaler())
])
# Categóricas: completar faltantes y convertir texto en números.
pipeline_categorico = Pipeline(steps=[
    ("imputador", SimpleImputer(strategy="most_frequent")),
    ("codificador", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

preprocesador = ColumnTransformer(transformers=[
    ("numericas", pipeline_numerico, variables_numericas),
    ("categoricas", pipeline_categorico, variables_categoricas)
])

# Aprender las transformaciones SOLO con entrenamiento.
X_train_procesado = preprocesador.fit_transform(X_train)
X_test_procesado = preprocesador.transform(X_test)

# TensorFlow trabaja con matrices numéricas.
X_train_procesado = np.asarray(
    X_train_procesado, dtype=np.float32
)
X_test_procesado = np.asarray(
    X_test_procesado, dtype=np.float32
)

y_train_np = y_train.to_numpy(dtype=np.float32)
y_test_np = y_test.to_numpy(dtype=np.int32)

print("\nForma de entrada para la red:", X_train_procesado.shape)
# 7. COMPENSAR EL DESEQUILIBRIO ENTRE CLASES
clases = np.unique(y_train_np.astype(int))

pesos = compute_class_weight(
    class_weight="balanced",
    classes=clases,
    y=y_train_np.astype(int)
)

pesos_clase = {
    int(clase): float(peso)
    for clase, peso in zip(clases, pesos)
}

print("\nPesos de las clases:", pesos_clase)
# 8. CONSTRUIR LA RED NEURONAL
modelo = tf.keras.Sequential([
    tf.keras.layers.Input(
        shape=(X_train_procesado.shape[1],)
    ),

    tf.keras.layers.Dense(
        32, activation="relu"
    ),

    tf.keras.layers.Dropout(0.25),

    tf.keras.layers.Dense(
        16, activation="relu"
    ),

    # Una salida entre 0 y 1.
    tf.keras.layers.Dense(
        1, activation="sigmoid"
    )
])

modelo.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy",
    metrics=[
        tf.keras.metrics.BinaryAccuracy(name="accuracy"),
        tf.keras.metrics.AUC(name="auc")
    ]
)

modelo.summary()
# 9. ENTRENAR EL MODELO
parada_temprana = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)

historial = modelo.fit(
    X_train_procesado,
    y_train_np,
    validation_split=0.20,
    epochs=150,
    batch_size=16,
    class_weight=pesos_clase,
    callbacks=[parada_temprana],
    verbose=1
)
# 10. PREDECIR CON LOS DATOS DE PRUEBA
# Probabilidad estimada de que Termd sea 1.
probabilidades = modelo.predict(
    X_test_procesado,
    verbose=0
).ravel()

# Umbral inicial de clasificación: 0.50.
predicciones = (probabilidades >= 0.50).astype(int)
# 11. EVALUAR EL MODELO
print("\nRESULTADOS")

print(
    "Accuracy:",
    round(accuracy_score(y_test_np, predicciones), 3)
)

print(
    "Precision:",
    round(
        precision_score(
            y_test_np, predicciones, zero_division=0
        ), 3
    )
)

print(
    "Recall:",
    round(
        recall_score(
            y_test_np, predicciones, zero_division=0
        ), 3
    )
)

print(
    "F1-score:",
    round(
        f1_score(
            y_test_np, predicciones, zero_division=0
        ), 3
    )
)

if len(np.unique(y_test_np)) == 2:
    print(
        "ROC-AUC:",
        round(roc_auc_score(y_test_np, probabilidades), 3)
    )

print("\nReporte de clasificación:")
print(classification_report(
    y_test_np,
    predicciones,
    labels=[0, 1],
    target_names=["No terminó", "Terminó"],
    zero_division=0
))
# 12. MATRIZ DE CONFUSIÓN
ConfusionMatrixDisplay(
    confusion_matrix=confusion_matrix(
        y_test_np, predicciones, labels=[0, 1]
    ),
    display_labels=["No terminó", "Terminó"]
).plot(cmap="Blues", values_format="d")

plt.title("TensorFlow: predicción de rotación")
plt.tight_layout()
plt.show()
# 13. VISUALIZAR EL APRENDIZAJE
fig, ejes = plt.subplots(1, 2, figsize=(12, 4))

ejes[0].plot(historial.history["loss"], label="Entrenamiento")
ejes[0].plot(historial.history["val_loss"], label="Validación")
ejes[0].set_title("Pérdida durante el entrenamiento")
ejes[0].set_xlabel("Época")
ejes[0].set_ylabel("Binary cross-entropy")
ejes[0].legend()

ejes[1].plot(
    historial.history["accuracy"], label="Entrenamiento"
)
ejes[1].plot(
    historial.history["val_accuracy"], label="Validación"
)
ejes[1].set_title("Exactitud durante el entrenamiento")
ejes[1].set_xlabel("Época")
ejes[1].set_ylabel("Accuracy")
ejes[1].legend()

plt.tight_layout()
plt.show()
# 14. MOSTRAR PREDICCIONES INDIVIDUALES
resultados = pd.DataFrame({
    "Termd_real": y_test_np,
    "Termd_predicho": predicciones,
    "Probabilidad_Termd_1": probabilidades
}, index=X_test.index)

resultados["Clasificacion"] = np.where(
    resultados["Termd_predicho"] == 1,
    "Terminó",
    "No terminó"
)

print("\nEjemplos de predicciones:")
print(resultados.head(10).round(3))
# 15. PROBAR CON UN EMPLEADO DEL CONJUNTO DE PRUEBA

empleado = X_test.iloc[[0]]
empleado_procesado = preprocesador.transform(empleado)

probabilidad = float(
    modelo.predict(empleado_procesado, verbose=0)[0, 0]
)

print("\nPREDICCIÓN INDIVIDUAL")
print("Probabilidad estimada de Termd = 1:",
      round(probabilidad, 3))

print(
    "Clase predicha:",
    1 if probabilidad >= 0.50 else 0
)
```
